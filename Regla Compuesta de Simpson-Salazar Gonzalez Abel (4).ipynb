Para comprobar el buen funcionamiento del codigo se usara como ejemplo la siguiente integral resuelta anteriormente en la tarea número 1

## $$I = \int_0^2 e^{2x}\sin(3x)\,dx$$

## SOLUCIÓN
Para poder resolver la integral mediante la regla compuesta de simpson debemos obtener h y n para la cual hacemos:

1) Obtenemos n
   
   Para esto usaremos la formula del error
   $$|E| \le \frac{b - a}{180} h^4 \max_{x \in [a,b]} |f^{(4)}(x)|$$

   Despues obtendremos h y el maximo de la cuarta derivada de nuestra función para susitutirlo en la formula del error

   1.1. Obteniendo h
     $$h = \frac{b - a}{n} = \frac{2}{n},$$
   
    1.2. Obtenemos la cuarta derivada de nuestra función
     #### Funcion inicial: $$f(x) = e^{2x} \sin(3x)$$
     #### Cuarta derivada: $$f^{(4)}(x) = -e^{2x} (119 \sin(3x) + 120 \cos(3x))$$
    1.3. Obteniendo el maximo de la cuarta derivada
     #### $$M = 9225.69$$
    1.4. Sustituyendo valores en la formua del error y acotando al error $10^{-4}$
     #### $$\frac{2}{180} \left(\frac{2}{n}\right)^4 (9225.69) \le 10^{-4}$$
    1.5. Despejando n tenemos
     #### $$n \ge 63.63$$
     #### $$n = 64$$
2) Sustituimos n en h
   $$h = \frac{2}{64} = \frac{1}{32} = 0.03125$$

Por lo tanto los valor de n y h son 
 $$n=64 , h=0.03125$$

```python
import math 
import sympy as sp 

#En esta parte del programa el codigo convierte a x como una incognita, la funcion ingresada desde el teclado la convierte a una funcion matematica la cual
#su variable es x y la convierte a una función f(x) 
x = sp.Symbol('x') 
expresion_str = input("Ingrese la función f(x) a integrar: ") 
expr = sp.sympify(expresion_str) 
f = sp.lambdify(x, expr, modules=['math'])
                
#Pedimos al que ejecuta el programa que ingrese los datos de la integral a evaluar (limites superiror e inferior de las integrales, 
#las particiones asi como la funcion a integrar)
a = float(input("Valor del limite inferior a: "))
b = float(input("Valor del límite superior b: "))
n = int(input("Valor de las particiones n: "))
print("El valor de a es: ", a)
print("El valor de b es: ", b)
print("El valor de n es: ", n)
#Paso 1: Obtenemos h
h = (b-a)/n
print("El valor de h es: ", h)

#Paso 2: Sumatorias iniciales
XI_0 = f(a) + f(b) 
XI_1 = 0 
XI_2 = 0
#Paso 3: Ciclo de iteraciones
for i in range(1, n):
#Paso 4: Calculo de X
  X = a + (i*h) 
#Paso 5: IF para i si es par o impar 
  if i % 2 == 0: 
   XI_2 = XI_2 + f(X) 
  else:
   XI_1 = XI_1 + f(X)
#Paso 6: Ecuacición final de la regla compuesta de simpson 
XI = h*(XI_0 + (2*XI_2) + (4*XI_1))/3 

#Declaracion de variables y codigo de color para resaltar el resultado final
NEGRITA = "\033[1m"
RESET = "\033[0m"
#Paso 7: Resultado
print(f"Resultado de la integración: {NEGRITA}{XI}{RESET}")



  
     

```

#### $$E = |-14.2141772311 - (-14.2139709396)| \times 100 =0.02062915 $$
